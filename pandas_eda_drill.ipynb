```python
import pandas as pd

data = {
    "Name": ["Riya", "Amit", "Neha", "Rahul", "Pooja"],
    "Age": [20, 21, 20, 22, 21],
    "Marks": [85, 72, 91, 68, 88],
    "City": ["Patan", "Mehsana", "Ahmedabad", "Patan", "Mehsana"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicates:")
print(df.duplicated().sum())

print("\nStatistics:")
print(df.describe())
```

```python
data_with_missing = {
    "Name": ["Riya", "Amit", "Neha", "Rahul", "Pooja"],
    "Age": [20, 21, None, 22, 21],
    "Marks": [85, 72, 91, 68, 88],
    "City": ["Patan", "Mehsana", "Ahmedabad", "Patan", "Mehsana"]
}

df_missing = pd.DataFrame(data_with_missing)

print(df_missing.isnull().sum())
```
